#Q7.4
#Explanation
When x is passed to the function, Python gives the function its value. The parameter val is a separate local variable inside the function.
 When we write:  val = val + 1

we are only changing the local variable val, not x.

After the function finishes, val disappears, and x is still 7.

If you wanted to change the value and keep it, you would need to return the new value:

```python
#Q7.5
def swap_message(val1, val2):
    # Swap local variables
    temp = val1
    val1 = val2
    val2 = temp

    print("Inside function:")
    print("val1 =", val1)
    print("val2 =", val2)

x = 6
y = 3

swap_message(x, y)

print("Outside function:")
print("x =", x)
print("y =", y)
```

#Q8.1
#Explain Error
Variables created inside a function are local variables. They can only be used inside that function.

In this code, val is a parameter of my_abs(), so it exists only while the function is running. When print(val) is executed outside the function, Python cannot find any variable named val, so it raises a NameError.

```python
#Q8.2
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```

#Q8.3
#Explain Error
The UnboundLocalError happens because when you try to modify counter inside the increment_counter() function, Python assumes it's a new local variable. Since you're trying to read its value (counter + 1) before assigning it a value within that local scope, it throws an error. To fix this, you need to explicitly tell Python that you want to modify the global counter by using the global keyword.

```python
#Q8.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
